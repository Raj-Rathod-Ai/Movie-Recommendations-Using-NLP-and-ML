```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("movies_metadata.csv", sep=',', engine='python', on_bad_lines='skip')
```

```python
df = df.drop_duplicates().reset_index(drop=True)
```

```python
df.columns
```

```python
df = df[['title','overview','genres','tagline','vote_average','popularity']]
```

```python
df
```

```python
df.isnull().sum()
```

```python
df = df.dropna(subset=['title'])
```

```python
df['overview']=df['overview'].fillna('')
```

```python
import ast
df['genres']=df['genres'].apply(lambda x: ' '.join([i['name'] for i in ast.literal_eval(x)]))
```

```python
df['tagline']=df['tagline'].fillna('')
```

```python
df['vote_average']=df['vote_average'].fillna(df['vote_average'].mean())
```

```python
df['popularity']=df['popularity'].fillna(df['popularity'].mode())
```

```python
df.isnull().sum()
```

```python
df['tags']=df['overview']+df['genres']+df['tagline']
```

```python
import nltk
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer
```

```python
nltk.download('stopwords')
nltk.download('wordnet')
```

```python
stop_words = set(stopwords.words('english'))
lemmatizer = WordNetLemmatizer()
```

```python
import re
def preprocess(text):
    text = str(text).lower()
    text = re.sub(r'[^\w\s]', ' ', text)
    words = text.split()
    words = [word for word in words if word not in stop_words]
    words = [lemmatizer.lemmatize(word) for word in words]
    return ' '.join(words)
```

```python
df['tags'] = df['tags'].apply(preprocess)
```

```python
df['tags'][0]
```

```python
df = df.reset_index(drop=True)
```

```python
indices = pd.Series(df.index, index=df['title']).drop_duplicates()
```

```python
indices
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
tfidf = TfidfVectorizer(max_features=500000,ngram_range=(1,2),stop_words='english')
```

```python
tfidf_matrix = tfidf.fit_transform(df['tags'])
```

```python
tfidf_matrix.shape
```

```python
from sklearn.metrics.pairwise import linear_kernel
```

```python
def recommend(title,n=10):
  if title not in indices:
    return ['Movie not found']
  idx = indices[title]
  sim_scores = linear_kernel(tfidf_matrix[idx],tfidf_matrix).flatten()
  similars_idx = sim_scores.argsort()[::-1][1:n+1]
  return df['title'].iloc[similars_idx]
```

```python
recommend('SpederMan')
```

```python
import pickle
pickle.dump(tfidf_matrix,open('tfidf_matrix.pkl','wb'))
pickle.dump(tfidf,open('tfidf.pkl','wb'))
df.to_pickle('df.pkl')
pickle.dump(indices,open('indices.pkl','wb'))
```
